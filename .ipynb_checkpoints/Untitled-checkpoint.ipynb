```python
python3 -m pip install nltk
import re
import sys

# ------------------------------------------------------------
# CONFIGURATION
# ------------------------------------------------------------

GROUP_NAME = "groupname"

INPUT_FILE = "cran.all"
OUTPUT_FILE = GROUP_NAME + "_processed.all"

# Put your stop words here or load them from a file.
# Example: stopwords.txt should contain one word per line.
STOPWORD_FILE = "stopwords.txt"


# ------------------------------------------------------------
# 1. TOKENIZATION
# ------------------------------------------------------------
#
# Input:
#     Text of a document
#
# Output:
#     List of tokens
#
# Tokenization separates the text into individual word-like
# strings.
#
# Example:
#
# "Flow of air, through a pipe."
#
# becomes:
#
# ["Flow", "of", "air", "through", "a", "pipe"]
#
# ------------------------------------------------------------

def tokenize(text):
    """
    Convert document text into a list of tokens.

    We consider sequences of alphabetic characters as tokens.
    """

    # Find sequences consisting of A-Z or a-z.
    tokens = re.findall(r"[A-Za-z]+", text)

    return tokens


# ------------------------------------------------------------
# 2. STEMMING
# ------------------------------------------------------------
#
# Convert each token into its stem using the Porter Stemmer.
#
# Example:
#
#     connected -> connect
#     connections -> connect
#     connecting -> connect
#
# The assignment explicitly allows using an existing Porter
# stemmer implementation.
#
# ------------------------------------------------------------

def stem(tokens, stemmer):
    """
    Apply Porter stemming to every token.
    """

    stemmed_tokens = []

    for token in tokens:
        stemmed_tokens.append(stemmer.stem(token))

    return stemmed_tokens


# ------------------------------------------------------------
# 3. STOP WORD REMOVAL
# ------------------------------------------------------------
#
# Remove common words that usually don't contribute much to
# information retrieval.
#
# Example:
#
# ["the", "flow", "of", "air"]
#
# becomes:
#
# ["flow", "air"]
#
# ------------------------------------------------------------

def remove_stopwords(tokens, stopwords):
    """
    Remove tokens that occur in the stop-word set.
    """

    result = []

    for token in tokens:
        if token not in stopwords:
            result.append(token)

    return result


# ------------------------------------------------------------
# 4. NORMALIZATION
# ------------------------------------------------------------
#
# Normalize tokens into a standard representation.
#
# We use lowercase normalization here.
#
# Example:
#
# ["Flow", "AIR", "Pressure"]
#
# becomes:
#
# ["flow", "air", "pressure"]
#
# ------------------------------------------------------------

def normalize(tokens):
    """
    Normalize tokens by converting them to lowercase.
    """

    return [token.lower() for token in tokens]


# ------------------------------------------------------------
# LOAD STOP WORDS
# ------------------------------------------------------------

def load_stopwords(filename):
    """
    Read stop words from a file.

    Expected format:

        the
        a
        an
        of
        in
        ...

    One stop word per line.
    """

    stopwords = set()

    try:
        with open(filename, "r", encoding="utf-8") as file:

            for line in file:
                word = line.strip().lower()

                if word:
                    stopwords.add(word)

    except FileNotFoundError:
        print("[ERROR] Stop-word file not found:", filename)
        sys.exit(1)

    print("[DEBUG] Loaded", len(stopwords), "stop words")

    return stopwords


# ------------------------------------------------------------
# PROCESS ONE DOCUMENT
# ------------------------------------------------------------

def process_document(text, stemmer, stopwords):
    """
    Perform the complete preprocessing pipeline:

        Tokenization
             ↓
        Normalization
             ↓
        Stop-word removal
             ↓
        Stemming

    Returns the final list of tokens.
    """

    # Step 1: Tokenization
    tokens = tokenize(text)

    print("[DEBUG] Tokens before processing:", len(tokens))

    # Step 2: Normalization
    tokens = normalize(tokens)

    # Step 3: Stop-word removal
    tokens = remove_stopwords(tokens, stopwords)

    print("[DEBUG] Tokens after stop-word removal:", len(tokens))

    # Step 4: Stemming
    tokens = stem(tokens, stemmer)

    print("[DEBUG] Final tokens:", len(tokens))

    return tokens


# ------------------------------------------------------------
# READ CRAN.ALL
# ------------------------------------------------------------

def preprocess_collection(input_file, output_file, stemmer, stopwords):
    """
    Read cran.all document by document.

    We only process:
        .I -> document ID
        .T -> title
        .W -> abstract

    The .A author section is ignored.
    """

    current_doc_id = None
    current_section = None

    # Store title + abstract of current document.
    current_text = []

    documents_processed = 0

    try:
        input_fp = open(input_file, "r", encoding="utf-8")
    except FileNotFoundError:
        print("[ERROR] Input file not found:", input_file)
        sys.exit(1)

    output_fp = open(output_file, "w", encoding="utf-8")

    print("[DEBUG] Starting preprocessing...")
    print("[DEBUG] Input :", input_file)
    print("[DEBUG] Output:", output_file)

    # --------------------------------------------------------
    # Process a document when we reach its next .I tag
    # --------------------------------------------------------

    def process_current_document():

        nonlocal current_doc_id
        nonlocal current_text
        nonlocal documents_processed

        if current_doc_id is None:
            return

        # Combine title and abstract.
        text = " ".join(current_text)

        print("\n[DEBUG] Processing document:", current_doc_id)

        # Run the four preprocessing steps.
        tokens = process_document(
            text,
            stemmer,
            stopwords
        )

        # Write document ID.
        output_fp.write(".I " + current_doc_id + "\n")

        # Write token tag.
        output_fp.write(".S\n")

        # Write all processed tokens.
        output_fp.write(" ".join(tokens) + "\n")

        documents_processed += 1

        # Reset document data.
        current_text = []

    # --------------------------------------------------------
    # Read cran.all line by line
    # --------------------------------------------------------

    for line in input_fp:

        line = line.rstrip("\n")

        # ----------------------------------------------------
        # .I = Document ID
        # ----------------------------------------------------

        if line.startswith(".I"):

            # First finish the previous document.
            process_current_document()

            # Example:
            #
            # .I 1
            #
            # split -> [".I", "1"]

            parts = line.split()

            if len(parts) >= 2:
                current_doc_id = parts[1]
            else:
                print("[WARNING] Invalid .I line:", line)
                current_doc_id = None

            current_section = "I"

        # ----------------------------------------------------
        # .T = Title
        # ----------------------------------------------------

        elif line.startswith(".T"):

            current_section = "T"

        # ----------------------------------------------------
        # .W = Abstract
        # ----------------------------------------------------

        elif line.startswith(".W"):

            current_section = "W"

        # ----------------------------------------------------
        # .A = Authors
        #
        # We ignore this section.
        # ----------------------------------------------------

        elif line.startswith(".A"):

            current_section = "A"

        # ----------------------------------------------------
        # Normal text line
        # ----------------------------------------------------

        else:

            # Only collect title and abstract.
            if current_section == "T" or current_section == "W":
                current_text.append(line)

    # Process final document.
    process_current_document()

    input_fp.close()
    output_fp.close()

    print("\n========================================")
    print("[DEBUG] PREPROCESSING COMPLETE")
    print("[DEBUG] Documents processed:", documents_processed)
    print("[DEBUG] Output file:", output_file)
    print("========================================")


# ------------------------------------------------------------
# MAIN
# ------------------------------------------------------------

def main():

    print("========================================")
    print(" Cranfield Text Preprocessor")
    print("========================================")

    # Import Porter Stemmer.
    #
    # If using NLTK's PorterStemmer:
    #
    # pip install nltk
    #
    # However, the assignment says the Porter stemmer may
    # be taken from any source.
    #
    try:
        from nltk.stem import PorterStemmer
    except ImportError:
        print("[ERROR] NLTK is not installed.")
        print("Install using:")
        print("    pip install nltk")
        sys.exit(1)

    stemmer = PorterStemmer()

    # Load stop words.
    stopwords = load_stopwords(STOPWORD_FILE)

    # Process entire Cranfield collection.
    preprocess_collection(
        INPUT_FILE,
        OUTPUT_FILE,
        stemmer,
        stopwords
    )


if __name__ == "__main__":
    main()
```
